```python
# Import required libraries
import sys
from pathlib import Path
import numpy as np

# Add parent directory to path
sys.path.insert(0, str(Path.cwd().parent))

from config.config import *
from src.models.model2 import BinaryClassificationTE
import src.utils as ut
from sklearn.metrics import accuracy_score, roc_auc_score
from persim import wasserstein
from gudhi import bottleneck_distance
import time

print("✓ Libraries imported successfully")
```

```python
# Load train, test datasets
train_dataset = ut.SeismicDataset(data_path=Path("../data/processed/signals_train.hdf5"))
test_dataset = ut.SeismicDataset(data_path=Path("../data/processed/signals_test.hdf5"))

X_train, y_train, mag_train = train_dataset.get_data()
X_test, y_test, mag_test = test_dataset.get_data()

print(f"\n✓ Datasets loaded successfully")
print(f"  Train: {len(train_dataset)} signals")
print(f"  Test: {len(test_dataset)} signals")
print(f"\nSample (signal, label, magnitude): {train_dataset[0]}")
```

```python
# Initialize model
model = BinaryClassificationTE(
    distance=wasserstein,
    weights=[2, 1],
    dim=20,
    tau=5,
    stride=1,
    maxdim=1,
    sample=10,
    thresh=np.inf,
    alpha=0.1
)

start_time = time.time()
# Fit model
model.fit(X_train, y_train)
end_time = time.time()
print(f"Training time: {(end_time - start_time)/60:.2f} minutes")

start_time = time.time()
# Evaluate model
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")
end_time = time.time()
print(f"Evaluation time: {(end_time - start_time)/60:.2f} minutes")
```
